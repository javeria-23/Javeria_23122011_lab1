```python
# AI Study Assistant Agent

student = {
    "name": "Student",
    "subjects": ["AI", "Math", "Programming"],
    "weak_topics": ["AI Agents"],
    "completed_topics": [],
    "quiz_score": 0
}

print("AI Study Assistant Started")
print("Student:", student["name"])

while True:

    print("\nChecking student progress...")

    # Check weak topics
    if len(student["weak_topics"]) > 0:
        topic = student["weak_topics"][0]

        print("Weak Topic:", topic)
        print("Recommendation: Study", topic)
        print("Generating practice questions...")

        student["completed_topics"].append(topic)
        student["weak_topics"].remove(topic)

    # Take a quiz
    print("\nQuiz Time!")
    score = int(input("Enter your quiz score (0-100): "))

    student["quiz_score"] = score

    if score < 60:
        print("Score is low.")
        print("Recommendation: Revise the topic and practice more.")

    elif score >= 60:
        print("Good performance!")
        print("Recommendation: Move to the next topic.")

    # Goal test
    if len(student["weak_topics"]) == 0 and score >= 60:
        print("\nStudy Goal Completed!")
        print("AI Study Assistant Session Finished.")
        break

    print("\nUpdating study plan...")
```
