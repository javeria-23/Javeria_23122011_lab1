```python
from google import genai

client = genai.Client(api_key=api_key)
```

```python
MODEL = "models/gemini-3.5-flash"
```

```python
university_info = {

    "university": {
        "name": "Federal Urdu University of Arts, Sciences & Technology",
        "short_name": "FUUAST",
        "campus": "Gulshan Campus"
    },

    "student": {
        "name": "Javeria",
        "program": "BS Computer Science",
        "semester": "6th Semester"
    },

    "department": {
        "name": "Department of Computer Science"
    },

    "courses": {

        "Artificial Intelligence": {
            "teacher": "Miss Uzma"
        },

        "Numerical and Symbolic Computing": {
            "teacher": "Sir Akhtar Raza"
        },

        "Natural Language Processing": {
            "teacher": "Sir Khalid"
        },

        "Professional Practices": {
            "teacher": "Miss Sana"
        },

        "Compiler Construction": {
            "teacher": "Miss Naheed"
        }
    }
}
```

```python
def university_chatbot(user_input):

    prompt = f"""
You are a helpful university chatbot for
Federal Urdu University of Arts, Sciences & Technology (FUUAST).

You are assisting a student named Javeria.

Student Information:
- Name: Javeria
- Program: BS Computer Science
- Semester: 6th Semester

Use the university information provided below to answer
university-related questions.

University Information:
{university_info}

Instructions:
- Answer clearly and politely.
- Use the dictionary information for university-specific questions.
- Do not make up university-specific information.
- If information is not available, say:
  "Sorry, this information is not available in my database."
- You may answer general educational questions using your knowledge.
- Keep the response easy for students to understand.

Student Question:
{user_input}

Chatbot Response:
"""

    response = client.models.generate_content(
        model=MODEL,
        contents=prompt
    )

    return response.text
```

```python
reply = university_chatbot("What is my name and which semester am I in?")

print(reply)
```

```python
reply = university_chatbot("Who teaches Artificial Intelligence?")

print(reply)
```

```python
reply = university_chatbot(
    "Who teaches Numerical and Symbolic Computing, NLP, "
    "Professional Practices and Compiler Construction?"
)

print(reply)
```

```python
print("FUUAST University Chatbot Started")
print("Type 'exit' to stop.\n")

while True:

    user = input("You: ")

    if user.lower() == "exit":
        print("Goodbye Javeria!")
        break

    try:

        reply = university_chatbot(user)

        print("\nFUUAST Chatbot:")
        print(reply)
        print()

    except Exception as e:

        print("Error:", e)
```
