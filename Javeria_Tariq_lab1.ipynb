```python
print("===== TRAVEL PLANNER =====")

destination = input("Where would you like to travel? ")
budget = int(input("Enter your available budget in PKR: "))

if destination.lower() == "naran":
    if budget >= 22000:
        print("Naran is a good choice for your budget!")
        print("Recommended: Lake Saif-ul-Malook, Babusar Top")
    else:
        print("Your budget is a little low for Naran.")

elif destination.lower() == "skardu":
    if budget >= 50000:
        print("Skardu is a good choice for your budget!")
        print("Recommended: Shangrila, Deosai, Upper Kachura Lake")
    else:
        print("Your budget is a little low for Skardu.")

elif destination.lower() == "murree":
    if budget >= 18000:
        print("Murree is a good choice for your budget!")
        print("Recommended: Mall Road, Patriata, Pindi Point")
    else:
        print("Your budget is a little low for Murree.")

else:
    print("Sorry! This destination is not available.")
    print("Available options: Naran, Skardu, Murree")

print("===== THANK YOU FOR USING TRAVEL PLANNER =====")
```
