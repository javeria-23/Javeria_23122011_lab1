```python
from google import genai

client = genai.Client(api_key="AQ.Ab8RN6I54V4IFDka3k6wYXEsTc164ctH6K7ZbQSCCgJF6K92rw")
```

```python
MODEL = "models/gemini-3.5-flash"
```

```python
def travel_agent(user_input):

    prompt = f"""
You are a professional travel agent for Pakistan tourism.

Provide a well-structured travel plan with clear headings:

1. Destination Overview
2. Budget Travel Options (Bus, Train, Flight with approx cost in PKR)
3. Places to Visit (specific locations)
4. Food Suggestions (local dishes)
5. Money-Saving Tips

Make response detailed but easy for students.

User request: {user_input}

Travel Plan:
"""

    response = client.models.generate_content(
        model=MODEL,
        contents=prompt
    )

    return response.text
```

```python
reply = travel_agent("I want to visit Lahore from Karachi")

print(reply)
```

```python
print("Gemini Travel Agent Started (type 'exit' to stop)\n")

while True:
    user = input("You: ")

    if user.lower() == "exit":
        print("Goodbye")
        break

    try:
        reply = travel_agent(user)
        print("\nAgent:\n", reply, "\n")

    except Exception as e:
        print("Error:", e)
```

```python
def study_assistant(user_input):

    prompt = f"""
You are an intelligent AI Study Assistant for students.

Analyze the student's request and provide a clear, helpful,
and easy-to-understand response.

The assistant should consider:
- Subject
- Topic
- Student's question
- Difficulty level
- Study time
- Important concepts
- Practice questions

Provide a well-structured response with clear headings:

1. Topic Overview
2. Simple Explanation
3. Important Points
4. Study Plan
5. Practice Questions
6. Quick Revision

Make the response detailed but easy for students.

Student request: {user_input}

AI Study Assistant Response:
"""

    response = client.models.generate_content(
        model=MODEL,
        contents=prompt
    )

    return response.text
```

```python
reply = study_assistant("Explain the basics of Python loops for a beginner student.")

print(reply)
```

```python
print("AI Study Assistant Started (type 'exit' to stop)\n")

while True:
    user = input("You: ")

    if user.lower() == "exit":
        print("Goodbye")
        break

    try:
        reply = study_assistant(user)
        print("\nAI Study Assistant:\n", reply, "\n")

    except Exception as e:
        print("Error:", e)
```
